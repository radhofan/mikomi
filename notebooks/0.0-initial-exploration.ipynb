# 0.0 Initial Data Exploration

Explore raw dataset and understand schema and inconsistencies.

```python
import json
from pathlib import Path
import pandas as pd

DATA_RAW = Path('../data/raw')
df_leads = pd.read_csv(DATA_RAW / 'leads_seed.csv')
print('Shape:', df_leads.shape)
df_leads.head()
```

## Column Info and Missing Values

```python
df_leads.info()
df_leads.isnull().sum()
```

## Lead Status and Source Values

```python
print(df_leads['Lead Status'].value_counts(dropna=False))
print(df_leads['Original Source'].value_counts(dropna=False))
```

## Website Form Submissions

```python
with open(DATA_RAW / 'website_form_submissions.json', 'r', encoding='utf-8') as f:
    submissions = json.load(f)

df_sub = pd.DataFrame(submissions)
print('Submissions count:', len(df_sub))
df_sub.head()
```
